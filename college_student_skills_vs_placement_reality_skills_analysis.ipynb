```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/college_students_skills_vs_placement_reality.csv')

plt.figure(figsize=(8,6))

sns.boxplot(
    data=df,
    x='placement_status',
    y='cgpa',
    hue='placement_status',
    legend=False
)

plt.title('CGPA Distribution by Placement Status')
plt.xlabel('Placement Status')
plt.ylabel('CGPA')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv('/content/college_students_skills_vs_placement_reality.csv')

plt.figure(figsize=(10,6))

sns.histplot(
    data=df[df['placement_status'] == 'Placed'],
    x='package_lpa',
    kde=True,
    bins=20
)

plt.title('Package Distribution of Placed Students')
plt.xlabel('Package (LPA)')
plt.ylabel('Frequency / Density')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("college_students_skills_vs_placement_reality.csv")

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="cgpa",
    bins=10,
    kde=True,
    color="purple",
    edgecolor="black"
)

plt.title("Distribution of Student CGPA", fontsize=16, fontweight="bold")
plt.xlabel("CGPA", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
sns.histplot(
    data=df,
    x="internships",
    bins=10,
    kde=True,
    color="orange",
    edgecolor="black"
)

plt.title("Distribution of Student Internships")
plt.xlabel("Number of Internships")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.violinplot(
    data=df,
    x='placement_status',
    y='cgpa',
    hue='placement_status',
    legend=False
)

plt.title('CGPA Distribution by Placement Status (Violin Plot)')
plt.xlabel('Placement Status')
plt.ylabel('CGPA')

plt.tight_layout()
plt.show()
```

```python
num_cols = ['cgpa', 'internships'] # Removed 'skills', 'projects_done', 'aptitude_score' as they were not found. Please verify correct column names if you wish to include them.

corr = df[num_cols].corr()

plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    fmt='.2f',
    square=True
)

plt.title('Correlation Matrix of Student Placement Factors')

plt.tight_layout()
plt.show()
```
